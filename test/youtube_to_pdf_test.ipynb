#### Tests download of a youtube video transcript, writing to pdf

```python
from youtube_transcript_api import YouTubeTranscriptApi
from fpdf import FPDF

# Extract video ID from the URL
video_url = "https://www.youtube.com/watch?v=TaG2zCVotxE"
video_id = video_url.split("v=")[1]

# Download the transcript
transcript = YouTubeTranscriptApi.get_transcript(video_id)

# Create a PDF object
pdf = FPDF()
pdf.add_page()
pdf.set_font("Arial", size=12)

# Write transcript to PDF
for entry in transcript:
    text = entry['text']
    pdf.multi_cell(0, 10, text)

# Save the PDF
outFNm = "tmp_yt_transcript.pdf"
pdf.output(outFNm)

print(f"Transcript downloaded and saved as {outFNm}")
```
