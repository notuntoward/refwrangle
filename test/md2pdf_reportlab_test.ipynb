```python
import pathlib as pl
from icecream import ic
import sys

refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of an .ipynb 
sys.path.append(str(refwrangle_dir))
import refwrangle as rfw
```

```python
bFNm = 'Ansolabehere24forecastTurnout.md'
input_file = rfw.attachments_as_md_cachedir / bFNm
output_file = pl.Path("tmp_output_from_memory.pdf")
```

```python
# Example usage
markdown_content = """
# Main Title

## Section 1
Content for section 1.

### Subsection 1.1
More detailed content.

## Section 2
Another main section.

### Subsection 2.1
Even more content here.
"""

ic(input_file, output_file)

with input_file.open('r', encoding='utf-8') as source_file:
    markdown_content = source_file.read()    

rfw.md2pdf_markdown(markdown_content, output_file, is_file=False)
```
