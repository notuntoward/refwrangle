#### Collections

```python
# all
zot.collections()[0:2]

# just top level collections
zot.collections_top()[0:2]
```

```python
# all names
[collection['data']['name'] for collection in zot.collections()][0:2]
```

```python
# getting the the key of a collection by the visible collection name
def get_collection_key(collection_name, zot):
    return [clctn for clctn in zot.collections() if clctn['data']['name'] == collection_name][0]['key']

politics_key = get_collection_key('Politics', zot)
politics_key
```

```python
# collections under a collection
[collection['data']['name'] for collection in zot.collections_sub(politics_key)]
```

```python
# items under a specific collection
zot.collection_items(politics_key)  # empty b/c there are no items immediately under Politics

pollMeth_key = get_collection_key('PollMethods', zot)
zot.collection_items(pollMeth_key)[0:2] # has item
```
