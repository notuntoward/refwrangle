#### Test ability to shift markdown headers hierarchically

```python
from icecream import ic
import pathlib as pl
import sys

refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of an .ipynb 
sys.path.append(str(refwrangle_dir))
import refwrangle as rfw

# Test cases
def run_tests():
    tests = [
        {
            "name": "Mixed header levels",
            "input": "## Level 2\nContent\n### Level 3\nMore content\n#### Level 4\nEven more content",
            "top_level": 1,
            "expected": "# Level 2\nContent\n## Level 3\nMore content\n### Level 4\nEven more content"
        },
        {
            "name": "All header levels",
            "input": "# L1\n## L2\n### L3\n#### L4\n##### L5\n###### L6\n####### L7",
            "top_level": 1,
            "expected": "# L1\n## L2\n### L3\n#### L4\n##### L5\n###### L6\nL7",  # fixed bogus L7
        },
        {
            "name": "Shift down and off",
            "input": "# L1\n## L2\n### L3\n#### L4\n##### L5\n###### L6\n####### L7",
            "top_level": 2,
            "expected": "## L1\n### L2\n#### L3\n##### L4\n###### L5\nL6\nL7",
        },
        {
            "name": "Single header",
            "input": "## Single header\nContent",
            "top_level": 1,
            "expected": "# Single header\nContent"
        },
        {
            "name": "No headers",
            "input": "Just plain text\nNo headers here",
            "top_level": 1,
            "expected": "Just plain text\nNo headers here"
        },
        {
            "name": "Top level 2",
            "input": "### Level 3\n#### Level 4\n##### Level 5",
            "top_level": 2,
            "expected": "## Level 3\n### Level 4\n#### Level 5"
        },
        {
            "name": "Top level None",
            "input": "## Level 2\n### Level 3\n#### Level 4",
            "top_level": None,
            "expected": "## Level 2\n### Level 3\n#### Level 4"
        },
        {
            "name": "Demote to text",
            "input": "#### Level 4\n##### Level 5\n###### Level 6\n####### Level 7",
            "top_level": 3,
            "expected": "### Level 4\n#### Level 5\n##### Level 6\nLevel 7"
        }
    ]

    for test in tests:
        result = rfw.shift_markdown_headers(test["input"], test["top_level"])
        if result == test["expected"]:
            print(f"Test '{test['name']}' passed.")
        else:
            print(f"Test '{test['name']}' failed.")
            print(f"Expected:\n{test['expected']}")
            print(f"Got:\n{result}")
        print()

    # Test for ValueError
    try:
        rfw.shift_markdown_headers("## Test", top_level=7)
        print("ValueError test failed: should have raised an exception")
    except ValueError:
        print("ValueError test passed: correctly raised an exception for invalid top_level")

# Run the tests
run_tests()
```
