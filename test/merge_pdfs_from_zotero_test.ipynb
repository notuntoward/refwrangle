### Get zotero item pdf attachments

Zotero DB pdf infoextraction done by load_zotero_data.ipynb

```python
from icecream import ic
import pathlib as pl
from pyzotero import zotero
import sys

refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of an .ipynb 
sys.path.append(str(refwrangle_dir))
import refwrangle as rfw
```

```python
zotinfo = rfw.load_pickle_data(rfw.extractedZoteroEntriesFNm) # all attachments from load_zotero_data.ipynb

# select which attachments to merge
pdf_attachments = zotinfo['attachment_files'].query('contentType == "application/pdf"')

merge_pdf_attachments = pdf_attachments[pdf_attachments['parentCollections'].apply(lambda x: 'PoliticalML' in x)]

# Standardize merged output separator page meta information names
info_renames = dict(Title='parentTitle',Author='parentFirstCreator',Source='parentVenue', Date='parentDate')
metaCols = merge_pdf_attachments[info_renames.values()].rename(info_renames, axis=1)

# Do the merge
pdfs_info = []
for ix, fullpath in merge_pdf_attachments.file_fullpath.items():
    pdfs_info.append({'file':str(fullpath), 'metainfo':metaCols.loc[ix].to_dict()})

rfw.merge_pdfs_with_structure(pdfs_info, "merged_articles.pdf")
print('Done.')
```
