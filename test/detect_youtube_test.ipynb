### Test detection of youtube videos

```python
import pathlib as pl
import sys

refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of your .ipynb 
sys.path.append(str(refwrangle_dir))
import refwrangle as rfw

from pyzotero import zotero
import pandas as pd
```

```python
# Retrieve all items from the Zotero library
zot = zotero.Zotero(rfw.library_id, 'user', rfw.api_key)
all_items = zot.everything(zot.top())
```

```python
# THING BEING TESTED
youtube_videos = [item for item in all_items if rfw.is_youtube_video(item)]

print(f"Total YouTube videos found: {len(youtube_videos)}")
ytinfo = []
for video in youtube_videos:
    ytinfo.append(pd.Series({k: video['data'][k] for k in ['title','url','date']}))

pd.DataFrame(ytinfo).set_index('date').sort_index()
```
