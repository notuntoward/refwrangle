```python
from litellm import get_model_list
import requests

def get_all_models():
    # Priority 1: Use LiteLLM’s maintained list
    try:
        return get_model_list(provider="perplexity")
    except:
        pass
    
    # Priority 2: Scrape Perplexity’s docs
    try:
        return get_perplexity_models()
    except:
        pass
    
    # Priority 3: Fallback to known models
    return ["sonar-pro", "llama-3.1-sonar-large-128k-online", "pplx-70b-online"]

models = get_all_models()

# Query all models with the same prompt
for model in models:
    response = requests.post(
        "https://api.perplexity.ai/chat/completions",
        headers={"Authorization": "Bearer YOUR_API_KEY"},
        json={
            "model": model,
            "messages": [{"role": "user", "content": "Your prompt here"}]
        }
    )
    print(f"Response from {model}:", response.json())
```
