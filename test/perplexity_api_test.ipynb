```python

messages = [
    {
        "role": "system",
        "content": (
            "You are an artificial intelligence assistant and you need to "
            "engage in a helpful, detailed, polite conversation with a user."
        ),
    },
    {   
        "role": "user",
        "content": (
            "How many stars are in the universe?"
        ),
    },
]

client = OpenAI(api_key=rfw.perplexity_api_key, base_url="https://api.perplexity.ai")
```

### Getting the complete response, but you must wait longer to get it

```python

# chat completion without streaming
response = client.chat.completions.create(
    model="sonar-pro",
    messages=messages,
)
#print(response)
```

```python
response_d = response.to_dict()
print(response_d['choices'][0]['message']['content'])
for ix, cite in enumerate(response_d['citations'], start=1):
    print(f'[{ix}] {cite}')
```

```python
for key, value in response.to_dict().items():
    print(f'{key=}\n{value}')
```

### Getting it incrementally, first chunk comes faster, like on the web UI

```python

# chat completion with streaming
response_stream = client.chat.completions.create(
    model="sonar-pro",
    messages=messages,
    stream=True,
)

responses = []
for response in response_stream:
    print(response)
    responses.append(response)
```

```python
output = []
for response in responses:
    output.append(response.to_dict()['choices'][0]['message']['content'])
    
output = " ".join(output)

print(output)
    
```
