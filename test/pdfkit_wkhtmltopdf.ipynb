```python
# The most basic: leavea a lot of cruft, no links

def convert_html_to_pdf(html_path, pdf_path, wkhtmltopdf_path):
    # Configure pdfkit to use the specified wkhtmltopdf installation
    config = pdfkit.configuration(wkhtmltopdf=str(wkhtmltopdf_path))

    # Read the HTML file
    with open(html_path, 'r', encoding='utf-8') as html_file:
        html_content = html_file.read()

    # Convert HTML to PDF
    try:
        pdfkit.from_string(html_content, str(pdf_path), configuration=config)
        print(f"PDF generated and saved at {pdf_path}")
    except Exception as e:
        print(f"PDF generation failed: {e}")


convert_html_to_pdf(html_file, pdf_file, wkhtmltopdf_path)
```
