```python
import pandas as pd
import re
from icecream import ic
```

```python
line = '#### I am a minister. I usually wake up at 4 a.m. My first half-hour is usually personal time, prayer, meditation, and then I start to engage with email and texts.'
#line = 'I am a minister. I usually wake up at 4 a.m. My first half-hour is usually personal time, prayer, meditation, and then I start to engage with email and texts.'
num_words = len(line.split())
ic(num_words)
ic(re.match(r'^#+\s', line) and not 2 <= num_words <= 13)

line.lstrip('#').strip()
```

```python

# From: https://www.perplexity.ai/search/when-i-use-the-python-library-QBEJf90rQfenHJiNQAIHxQ
#words_per_page = 500 # academic
#words_per_heading_max = 


def count_headings_and_words(markdown_content):
    heading_counts = []
    total_words = 0
    heading_count = 0

    for line in markdown_content.splitlines():
        stripped_line = line.strip()
        if stripped_line.startswith('#'):
            heading_count += 1
            # Count words in heading (excluding the '#' characters)
            words_in_heading = len(stripped_line.lstrip('#').strip().split())
            heading_counts.append(words_in_heading)
        
        # Count total words
        total_words += len(line.split())

    # Calculate heading word count statistics
    min_heading_words = min(heading_counts) if heading_counts else 0
    max_heading_words = max(heading_counts) if heading_counts else 0
    avg_heading_words = sum(heading_counts) / len(heading_counts) if heading_counts else 0

    return {
        'total_headings': heading_count,
        'total_words': total_words,
        'min_heading_words': min_heading_words,
        'max_heading_words': max_heading_words,
        'avg_heading_words': avg_heading_words
    }

# Example usage
markdown_content = """
# This is a Long Heading with Many Words

Some content here.

## Shorter Heading

More content.

### Very Short

Even more content here.

#### Another Heading with Several Words in It

Final paragraph.
"""

results = count_headings_and_words(markdown_content)

print(f"Number of headings: {results['total_headings']}")
print(f"Total number of words: {results['total_words']}")
print(f"Minimum words in a heading: {results['min_heading_words']}")
print(f"Maximum words in a heading: {results['max_heading_words']}")
print(f"Average words in headings: {results['avg_heading_words']:.2f}")


def remove_weird_len_headers(markdown_text):
    """Remove out-of-range length headers."""

    lines = markdown_text.split('\n')
    corrected_lines = []
    for line in lines:
        num_words = len(line.split())
        isHeading = re.match(r'^#+\s', line)
        ic(isHeading, num_words)
        if re.match(r'^#+\s', line) and not 3 <= num_words <= 13:
        #if isHeading and not (3 <= num_words <= 13):            
            ln_corr = line.lstrip('#').strip()
            ic('corrected.', line, ln_corr)
            corrected_lines.append(ln_corr)
        else:
            corrected_lines.append(line)
    return '\n'.join(corrected_lines)


test_md = """
Lui
39,
bira
Dem
une


-----

Moderator, Kristen Soltis Anderson
## When you wake up, what does the first hour of your day look like?

Matthew, 43, Tennessee, white, Republican, attorney

Helping get my three kids ready. Then work email.

Michael, 52, Illinois, white, Republican, minister

#### I am a minister. I usually wake up at 4 a.m. My first half-hour is usually personal time, prayer, meditation, and then I start to engage with email and texts.

Ron W., 39, California, Black, independent, construction

#### I wake up at 5 a.m., and I’ll work out for an hour.

Moderator, Kristen Soltis Anderson
Now I want to hear about the last hour of your day.

Rich, 54, New York, Latino, Democrat, construction

#### I get in bed, get on my laptop, answer some emails. Have the TV on in the background, watch some news and try to decompress from the day.

Brett, 62, California, Black, Democrat, retired

Almost every night, I have some ice cream, watch a

# Heading with OK length
# Sh

alskfjal;sdjf

# Another OK length

-----

## I want to hear more about how you get news and information about important issues affecting the country.

Ken, 39, western U.S., Pacific Islander, Democrat, teacher

#### Online. I use MSN to filter and then look at other news outlets. I’ll listen to Joe Rogan’s podcast, but at the same time, I’ll listen to something else from another side. I try to be nonbiased, because it’s so easy to look at something in just one way, but that’s just half the story.

Jon, 45, Pennsylvania, white, Republican, business owner

#### I get most of my info from X. That’s a constant stream of intel throughout the day.

Moderator, Kristen Soltis Anderson
Are there particular accounts that you follow on X?

Jon, 45, Pennsylvania, white, Republican, business owner

#### Folks on the right, folks on the left, podcasters I enjoy. Obviously we all get bombarded with Elon Musk posts. If I am spending that last hour of the day with my wife, we’ll watch some of our favorite YouTube news feeds on our TV. The one we watch


-----

#### I’ll usually go online between CNN and Fox News and see what I think is right, what I think is wrong and what I agree or don’t agree with.

Luis, 39, Florida, biracial, Democrat, unemployed

#### I go to Reddit. Usually you can see both sides because somebody will point out the opposite side of whatever the issue is.


-----

Michael, 52, Illinois, white, Republican, minister

#### I use Newsmax a lot, but I also use Turning Point USA, CBN and i24news, which is Israel’s 24-hour"""

hdat_before = count_headings_and_words(test_md)
hdat_after = count_headings_and_words(remove_weird_len_headers(test_md))

pd.DataFrame([hdat_before, hdat_after], index=['before','after'])
```
