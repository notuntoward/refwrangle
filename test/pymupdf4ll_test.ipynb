```python
import pymupdf4llm
import pathlib as pl

refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of your .ipynb 
test_dir = refwrangle_dir / 'test'

# a paper's pdf
pdf_file = test_dir / "Nizharadze23PredictingGapDayAhead.pdf"
# pypeteer test output
pdf_file = refwrangle_dir / "tmp_pypeteer_out.pdf"
# wkhtmltopdf test
pdf_file = test_dir /  'tmp_wkhtmltopdf.pdf'
# default playwright test output (playwright_test)
pdf_file = test_dir /  'tmp_playwright.pdf'

md_file = test_dir / "tmp_pymupdf4llm_out.md"

md_text = pymupdf4llm.to_markdown(pdf_file)

md_file.write_bytes(md_text.encode())

print(f"Done writing to {md_file}.")
```
