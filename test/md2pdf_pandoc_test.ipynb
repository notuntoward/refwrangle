### Test Markdown files to pdf files using pandoc

Seems to work.  But I remember seeing encoding problems elsewhere.

```python
import pathlib as pl
from icecream import ic
import sys

refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of an .ipynb 
sys.path.append(str(refwrangle_dir))
import refwrangle as rfw

bFNm = 'Bloomberg23politAdsVirtReality.md'
input_file = rfw.attachments_as_md_cachedir / bFNm
output_file = pl.Path("tmp_md2pdf_pandoc.pdf")
ic(input_file, output_file)
```

```python
import subprocess

def convert_md_to_pdf(input_file, output_file):
    command = f"pandoc {input_file} -o {output_file} --toc --toc-depth=3"
    subprocess.run(command, shell=True, check=True)

convert_md_to_pdf(input_file, output_file)
```
