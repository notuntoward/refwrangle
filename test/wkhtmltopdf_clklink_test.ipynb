```python
# works, but links aren't clickable or vertically centered.

import subprocess
import pathlib as pl
from icecream import ic

wkhtmltopdf_exe = pl.Path(r"C:\Program Files\wkhtmltopdf\bin\wkhtmltopdf.exe")
hpath = pl.Path(r'C:/Users/scott/OneDrive/share/ref/obsidian/Obsidian Share Vault/lit/lit_sources/')

html_file_path = hpath / 'Tumulty24FrischLearnedDemsShould.html'  # works, but complex WA post page has tons of extraneous link junk
# hpath / 'Walther24barstoolConservatism.html'     # works
#html_file_path = hpath / 'Yan24berkeleyFuncCallLeaderBrd.html'     # works now, used to fail on all b/c .html was corrupt

output_pdf_path = pl.Path("./tmp_wkhtmltopdf.pdf").resolve()

def create_header_html(input_file):
    basename = input_file.stem
    link = input_file.as_uri()
    header_html_content = f"""
    <!DOCTYPE HTML>
    <html>
    <head>
        <style>
            body {{
                font-family: Arial, sans-serif;
                font-size: 16px; /* Increase font size */
                margin: 0;
                padding: 0;
                display: flex;
                justify-content: center;
                align-items: center;
                height: 100%;
            }}
            a {{
                text-decoration: none;
                color: blue;
            }}
        </style>
    </head>
    <body>
        <a href="{link}">{basename}</a>
    </body>
    </html>
    """
    header_html_path = pl.Path("./header.html").resolve()
    with open(header_html_path, "w") as f:
        f.write(header_html_content)
    return header_html_path

def convert_html_to_pdf(input_file, output_file):
    ic(input_file, output_file)
    
    # Create header HTML with link
    header_html_path = create_header_html(input_file)
    
    # Run wkhtmltopdf with header-html option and proper margins
    subprocess.run([
        str(wkhtmltopdf_exe),
        "--header-html", str(header_html_path.as_uri()),
        "--margin-top", "20mm",
        "--enable-local-file-access",
        str(input_file),
        str(output_file)
    ])

convert_html_to_pdf(html_file_path.resolve(), output_pdf_path)
```
