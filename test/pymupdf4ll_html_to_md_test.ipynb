### Test of html --> markdown (playwright or wkhtmltopdf and pymupdf4llm)

The hope was that, with relatively faithful html -> pdf, then pymupdf4llm would be able to use its AI to create clean markdown.  

- Playwright is fairly good at html --> pdf but pymupdf4llm can't figure out what's junk, probably b/c it's trained on pdfs from articles, etc., rather than web pages.  Sometimes, though the markdown is reasonably good, but far from perfect.  clear problem w/ breaking single sentences into multiple paragraphs, based on line endings in multiple line sentence.
- wkhtmltopdf for html --> pdf removes more junk, but pymupdf4llm seems more confused by what's left over.  Also has line ending problem.
- [The PDF Extraction Revolution: Why PymuPDF4llm is Your New Best Friend (and LlamaParse is Crying)](https://ai.gopubby.com/the-pdf-extraction-revolution-why-pymupdf4llm-is-your-new-best-friend-and-llamaparse-is-crying-e57882dee7f8)
