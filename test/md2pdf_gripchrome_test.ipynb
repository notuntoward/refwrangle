```python
from pathlib import Path
import subprocess
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time

def markdown_to_pdf(markdown_file: Path, output_pdf: Path):
    # Ensure input and output are Path objects
    markdown_file = Path(markdown_file)
    output_pdf = Path(output_pdf)

    # Install Grip if not already installed
    subprocess.run(['pip', 'install', 'grip'], check=True)

    # Use Grip to render Markdown as HTML
    html_file = markdown_file.with_suffix('.html')
    subprocess.run(['grip', str(markdown_file), '--export', str(html_file)], check=True)

    # Set up Chrome options for headless PDF printing
    chrome_options = Options()
    chrome_options.add_argument('--headless')
    chrome_options.add_argument('--disable-gpu')
    chrome_options.add_argument('--no-sandbox')
    chrome_options.add_argument(f'--print-to-pdf={output_pdf.resolve()}')

    # Set up the Chrome service using WebDriver Manager
    service = Service(ChromeDriverManager().install())

    # Initialize the Chrome driver with the service and options
    driver = webdriver.Chrome(service=service, options=chrome_options)

    # Load the HTML file
    driver.get(html_file.resolve().as_uri())

    # Wait for the page to load (adjust the sleep time if needed)
    time.sleep(5)

    # Close the browser
    driver.quit()

    # Remove the temporary HTML file
    html_file.unlink()

    print(f"PDF saved as {output_pdf}")

# # Example usage
# input_file = Path('README.md')
# output_file = Path('output.pdf')
# markdown_to_pdf(input_file, output_file)
```
