```python
import trafilatura
import pathlib as pl
import sys
refwrangle_dir = pl.Path('~/ref/refwrangle').expanduser() # can't reliably get dir of your .ipynb 
sys.path.append(str(refwrangle_dir))
import refwrangle as rfw
from icecream import ic

html_to_pdf_lib = 'playwright'
#html_to_pdf_lib = 'whktmltopdf'

#hbasename = 'Klein24itsCorruptionStupid.html' 
#hbasename = 'ODonnel24newsEngageTrumpWin.html' 
hbasename = 'Tumulty24FrischLearnedDemsShould.html' # success=0, even with prefer_recall=True
#hbasename = 'Walther24barstoolConservatism.html'  
#hbasename = 'Yan24berkeleyFuncCallLeaderBrd.html'  # misses a lot of content, even with prefer_recall=True
html_file = rfw.lit_attachment_dir_shared / hbasename

test_dir = refwrangle_dir / 'test'
md_file = test_dir / 'tmp_trafilatura_html_to_md.md'

from trafilatura import extract, extract_metadata
from pathlib import Path
import json
from datetime import datetime

def html_to_markdown_with_metadata(input_html_path, output_markdown_path):
    """
    Not sure I need this....  
    Convert HTML file to markdown with metadata
    
    Args:
        input_html_path: Path to input HTML file
        output_markdown_path: Path to output markdown file
    """
    # Read HTML file
    with open(input_html_path, 'r', encoding='utf-8') as f:
        html_content = f.read()
    
    # Extract metadata
    metadata = extract_metadata(html_content)
    
    # Extract content with specified options
    markdown_content = extract(
        html_content,
        output_format='markdown',
        include_comments=False,
        include_tables=True,
        include_formatting=True,
        include_links=True,
        include_images=False,
        favor_recall=True  # misses worse than extra junk?
    )
    
    if markdown_content:
        # Create markdown frontmatter
        frontmatter = {
            'title': metadata.title,
            'author': metadata.author,
            'url': metadata.url,
            'hostname': metadata.hostname,
            'description': metadata.description,
            'sitename': metadata.sitename,
            'date': metadata.date,
            'categories': metadata.categories,
            'tags': metadata.tags,
            'extracted_date': datetime.now().isoformat()
        }
        
        # Remove None values from frontmatter
        frontmatter = {k: v for k, v in frontmatter.items() if v is not None}
        
        # Combine frontmatter and content
        full_content = (
            "---\n"
            f"{json.dumps(frontmatter, indent=2, ensure_ascii=False)}\n"
            "---\n\n"
            f"{markdown_content}"
        )
        
        # Write markdown output
        with open(output_markdown_path, 'w', encoding='utf-8') as f:
            f.write(full_content)
        return True
    return False

# Example usage for single file
def convert_file(html_file):
    input_path = Path(html_file)
    output_path = input_path.with_suffix('.md')
    success = html_to_markdown_with_metadata(input_path, output_path)
    print(f"Conversion {'successful' if success else 'failed'} for {input_path}")

print(f'Reading {html_file}')
success = html_to_markdown(html_file, md_file)

print(f"Done writing to {md_file}.  {success=}")
```
