#### Test wkhtmltopdf commandline html --> pdf

It doesn't clean, but does a good job with html->pdf, although where BS and trafilatura crash, it produces a blank page, which actually was more robust, as that html file on this page was corrupted.  Although a binary must be installed, it requires less config that beautiful soup or weaspyprint (GTK+ and fonts config to avoid error messages).

Note that the lib `pdfkit` is an interface to this, giving more control (see [perplexity](https://www.perplexity.ai/search/write-the-python-to-read-and-h-G5A__52rSJ2f378TtNdn7A#0]))
